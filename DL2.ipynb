```python
import numpy as np
#for AND gate (Input/Output)
x_list = np.array([[0,0],[0,1],[1,0],[1,1]])
y_list = np.array([0,0,0,1])
epochs = 4
w,b = np.zeros(2),0
for _ in range(epochs):
  for x,y in zip(x_list,y_list):
    z = np.dot(x,w) + b
    #condition area for y_pred
    y_pred = 1 if z >= 0 else 0
    E = y - y_pred
    print("error : ",E,"<- at epoch",_)
    w += E*x
    b += E
    print(w,b)
```

```python
#Add Logic to stop the training when the error is "ZERO" for all the inputs

import numpy as np
#for AND gate (Input/Output)
x_list = np.array([[0,0],[0,1],[1,0],[1,1]])
y_list = np.array([0,0,0,1])
epochs = 10
w,b = 2 * np.random.rand(2) - 1, 2 * np.random.rand() - 1
for epoch in range(epochs):
  all_correct = True
  for x,y in zip(x_list,y_list):
    z = np.dot(x,w) + b
    #condition area for y_pred
    y_pred = 1 if z >= 0 else 0
    E = y - y_pred
    print("error : ",E,"<- at epoch",epoch)
    if E != 0:
      all_correct = False
      w += E*x
      b += E
    print(w,b)
  if all_correct:
    print(f"Training stopped at epoch {epoch} as all errors are zero.")
    break
```

```python
#Repeat the same process for OR gate

#For OR gate (Input/Output)
x_list_or = np.array([[0,0],[0,1],[1,0],[1,1]])
y_list_or = np.array([0,1,1,1]) #OR gate outputs
epochs_or = 10
w_or, b_or = 2 * np.random.rand(2) - 1, 2 * np.random.rand() - 1
print("Training Perceptron for OR Gate:")
for epoch in range(epochs_or):
  all_correct = True
  for x,y in zip(x_list_or,y_list_or):
    z = np.dot(x,w_or) + b_or
    y_pred = 1 if z >= 0 else 0
    E = y - y_pred
    print(f"  Epoch {epoch}, Input: {x}, Target: {y}, Predicted: {y_pred}, Error: {E}")
    if E != 0:
      all_correct = False
      w_or += E*x
      b_or += E
    print(f"  Weights: {w_or}, Bias: {b_or}")
  if all_correct:
    print(f"Training stopped at epoch {epoch} as all errors are zero for OR gate.")
    break
print(f"\nFinal weights for OR gate: {w_or}")
print(f"Final bias for OR gate: {b_or}")
```

```python
#Try the same process for XOR gate and show that the network will not stabilize (Converge)

# For XOR gate (Input/Output)
import numpy as np
x_list_xor = np.array([[0,0],[0,1],[1,0],[1,1]])
y_list_xor = np.array([0,1,1,0]) # XOR gate outputs
epochs_xor = 20 # Increase epochs to show non-convergence more clearly
w_xor, b_xor = np.random.rand(2), np.random.rand()
print("Training Perceptron for XOR Gate:")
for epoch in range(epochs_xor):
  all_correct = True
  for x,y in zip(x_list_xor,y_list_xor):
    z = np.dot(x,w_xor) + b_xor
    y_pred = 1 if z >= 0 else 0
    E = y - y_pred
    print(f"  Epoch {epoch}, Input: {x}, Target: {y}, Predicted: {y_pred}, Error: {E}")
    if E != 0:
      all_correct = False
      w_xor += E*x
      b_xor += E
    print(f"  Weights: {w_xor}, Bias: {b_xor}")
  if all_correct:
    print(f"Training stopped at epoch {epoch} as all errors are zero for XOR gate.")
    break
print(f"\nFinal weights for XOR gate: {w_xor}")
print(f"Final bias for XOR gate: {b_xor}")
if not all_correct:
       print("\nAs expected, the perceptron did not converge for the XOR gate. This is because XOR is not linearly separable.")
```
