```python
import numpy as np
def mc_pitts_neuron(x, w, theta):
    return 1 if np.dot(x, w) >= theta else 0
# AND
print("\nAND Gate:")
w_and = np.array([1, 1])
theta_and = 2
for x_val in [(0, 0), (0, 1), (1, 0), (1, 1)]:
    print(f"Input: {x_val}, Output: {mc_pitts_neuron(np.array(x_val), w_and, theta_and)}")

# OR Gate
print("\nOR Gate:")
w_or = np.array([1, 1])
theta_or = 1
for x_val in [(0, 0), (0, 1), (1, 0), (1, 1)]:
    print(f"Input: {x_val}, Output: {mc_pitts_neuron(np.array(x_val), w_or, theta_or)}")

print("\nAnish Kumar | 22 Jan")
```

```python
import numpy as np
def art_neuron(x, w, b, activation='step'):
    z = np.dot(x, w) + b
    if activation == 'step':
        return 1 if z >= 0 else 0
    elif activation == 'sigmoid':
        return 1 / (1 + np.exp(-z))
    elif activation == 'relu':
        return max(0, z)
    else:
        return "Invalid activation function"
x = np.array([0, 0])
w = np.array([0, 0])
b = 0
print("step output:", art_neuron(x, w, b, 'step'))
print("sigmoid output:", art_neuron(x, w, b, 'sigmoid'))
print("relu output:", art_neuron(x, w, b, 'relu'))
print("\nAnish Kumar | 22 Jan")
```
